###  3. Символічні обчислення та алгебра виразів
а) оголосіть змінні $x=10$ і $y=15$ та обчисліть їхню суму і добуток;<br>
б) оголосіть символічну змінну $z$ і обчисліть вираз $3z^2 + 2z - 5$;<br>
в) спростіть вираз $2x^2 + 4x - 2x^2 + x$;<br>
г) розкрийте дужки у виразі $(x + 2)(x - 3)$;<br>
д) розкладіть на множники тричлен $x^2 - 9x + 20$;<br>
е) розв'яжіть рівняння $x^2 - 4x + 3 = 0$.<br>

```sage
x = 10
y = 15
suma = x + y
dobutok = x * y
print(f"а) Сума: {suma}, Добуток: {dobutok}")

var('z')
vyraz_b = 3*z^2 + 2*z - 5
print(f"б) Вираз для z: {vyraz_b}")

var('x')
vyraz_v = 2*x^2 + 4*x - 2*x^2 + x
sprosheno = vyraz_v.simplify_full()
print(f"в) Спрощений вираз: {sprosheno}")

vyraz_g = (x + 2)*(x - 3)
rozkryto = vyraz_g.expand()
print(f"г) Після розкриття дужок: {rozkryto}")

trychlen = x^2 - 9*x + 20
mnozhnyky = trychlen.factor()
print(f"д) Розклад на множники: {mnozhnyky}")

rivnyannya = x^2 - 4*x + 3 == 0
rozwiazok = solve(rivnyannya, x)
print(f"е) Розв'язок рівняння: {rozwiazok}")
```

### 4. Похідні, інтеграли та графіки функцій

а) побудуйте графік функції $f(x) = x^2 - 4x + 3$ на відрізку $[-5, 5]$; <br>
б) знайдіть похідну функції $f(x) = x^3 + 5x^2 - 2x + 7$; <br>
в) знайдіть невизначений інтеграл функції $2x^3 - x^2 + 5$; <br>
г) обчисліть визначений інтеграл $\int_{1}^{4} x^2\, dx$; <br>
д) оголосіть функцію $f(x) = 3x^2 - 2x + 1$ та обчисліть $f(5)$; <br>
е) побудуйте на одній площині графіки $f(x)=x^2$ і $g(x)=-x^2+5$ на $[-3,3]$.


```sage
print("а: ")
x = var('x')
f(x) = x^2 - 4*x + 3
plot(f(x), (x, -5, 5), axes_labels=['x', 'y'], title="Графік f(x) = x^2 - 4x + 3")
```

```sage
print("б: ")
x = var('x')
f(x) = x^3 + 5*x^2 - 2*x + 7
df = diff(f(x), x)
show(df)
```

```sage
print("в: ")
x = var('x')
f(x) = 2*x^3 - x^2 + 5
F = integral(f(x), x)
show(F)
```

```sage
print("г: ")
x = var('x')
res = integral(x^2, x, 1, 4)
show(res)
```

```sage
print("д: ")
x = var('x')
f(x) = 3*x^2 - 2*x + 1
print(f(5))
```

```sage
print("е: ")
x = var('x')
p1 = plot(x^2, (x, -3, 3), color='blue', legend_label='f(x) = x^2')
p2 = plot(-x^2 + 5, (x, -3, 3), color='red', legend_label='g(x) = -x^2 + 5')
show(p1 + p2, axes_labels=['x', 'y'], title="Суміщений графік")
```

### 5. Програмування
а) напишіть умовний оператор, який визначає, чи число $n$ додатне, від'ємне або дорівнює нулю; <br>
б) напишіть цикл, що виводить числа від 1 до 10; <br>
в) напишіть функцію, яка перевіряє чи є задане число $n$ простим.

```sage
print("а: ")
n = int(input())
if n > 0:
    print("n - додатнє")
elif n == 0:
    print("n = 0")
else:
    print("n - від'ємне")
```

```sage
print("б: ")
for i in range(1, 11):
    print(i, end =" ")
```

```sage
print("в: ")
def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6     
    return True
n = int(input())
print(is_prime(n))
```
